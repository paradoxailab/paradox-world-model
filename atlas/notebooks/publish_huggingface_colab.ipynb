# Publish every ParadoxAI Lab model to Hugging Face

Runtime: **CPU** is enough. Before running: create a Hugging Face **write** token (huggingface.co/settings/tokens) and add it in Colab's **Secrets** panel (key icon, left) as `HF_TOKEN`, with notebook access on. The token is read from the secret store and never printed.

Then *Runtime → Run all*. The last cell prints `PUBLISHED 56 models (all SHA-256 verified)` and the collection link.

```
import os
from google.colab import userdata
os.environ['HF_TOKEN'] = userdata.get('HF_TOKEN')
print('HF_TOKEN loaded from Colab secrets:', bool(os.environ['HF_TOKEN']))
```

```
from huggingface_hub import HfApi
import json, subprocess
me = HfApi().whoami()['name']
plan = json.load(open('models/hf/plan.json'))
print('Hugging Face account:', me, '| cards written for:', plan['ns'])
if me != plan['ns']:
    subprocess.run(['python', 'models/hf/make_hf_cards.py', '--ns', me], check=True)
```
